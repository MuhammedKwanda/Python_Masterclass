> Real-world data is messy. Cleaning it is 70% of the work.

```
import pandas as pd
import numpy as np
data = {
    "Name":  ["Amina","Bello",None,"Dauda","Emeka","Amina","George"],
    "Age":   [20, 22, 21, None, 200, 20, 24],
    "Score": [92.0,78.0,85.0,67.0,None,92.0,72.0],
    "City":  ["Lagos","abuja","KANO"," Lagos","Enugu","Lagos",None]
}
df = pd.DataFrame(data)
print(df)
```

```
print(df.isnull().sum())
print(f"Total missing: {df.isnull().sum().sum()}")
```

```
df2 = df.copy()
df2.dropna(subset=["Name"], inplace=True)
df2["Score"].fillna(df2["Score"].median(), inplace=True)
df2["Age"].fillna(df2["Age"].mean(), inplace=True)
df2["City"].fillna("Unknown", inplace=True)
print(df2)
```

```
df2.drop_duplicates(inplace=True)
df2["Age"] = df2["Age"].astype(int)
df2["City"] = df2["City"].str.strip().str.title()
df2 = df2[df2["Age"] <= 100]
print("Clean data:")
print(df2)
```
