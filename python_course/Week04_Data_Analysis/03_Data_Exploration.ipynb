```
import pandas as pd
import numpy as np
np.random.seed(42)
n = 100
df = pd.DataFrame({
    "Name": [f"Student_{i}" for i in range(n)],
    "Age": np.random.randint(18,30,n),
    "Gender": np.random.choice(["Male","Female"],n),
    "Math": np.random.randint(40,100,n),
    "Science": np.random.randint(40,100,n),
    "English": np.random.randint(40,100,n),
    "City": np.random.choice(["Lagos","Abuja","Kano","Enugu","Kaduna"],n)
})
print(df.shape)
```

```
print(df.head())
df.info()
print(df.describe())
print(df["City"].value_counts())
```

```
print(df[["Math","Science","English"]].corr())
print(df.sort_values("Math", ascending=False).head(5)[["Name","Math","City"]])
```
