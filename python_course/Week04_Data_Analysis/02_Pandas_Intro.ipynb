## Series & DataFrames

```
import pandas as pd
import numpy as np
print(f"Pandas: {pd.__version__}")
```

```
s = pd.Series([85,92,78,95,67], index=["Amina","Bello","Chioma","Dauda","Emeka"])
print(s)
print(s["Bello"])
print(s[s > 80])
```

```
data = {
    "Name": ["Amina","Bello","Chioma","Dauda","Emeka"],
    "Age":  [20, 22, 21, 23, 20],
    "Score":[92, 78, 85, 67, 88],
    "City": ["Lagos","Abuja","Kano","Lagos","Abuja"]
}
df = pd.DataFrame(data)
print(df)
print(f"Shape: {df.shape}")
```

```
print(df.head(3))
print(df.tail(2))
df.info()
print(df.describe())
print(df.dtypes)
```

```
print(df["Score"])
print(df[["Name","Score"]].head())
print(df.iloc[0])
print(df.sort_values("Score", ascending=False))
```
