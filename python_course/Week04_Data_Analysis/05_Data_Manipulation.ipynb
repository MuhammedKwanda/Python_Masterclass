```
import pandas as pd
import numpy as np
np.random.seed(42)
n = 50
df = pd.DataFrame({
    "Name": [f"Student_{i}" for i in range(n)],
    "Gender": np.random.choice(["Male","Female"],n),
    "City": np.random.choice(["Lagos","Abuja","Kano","Enugu"],n),
    "Department": np.random.choice(["CS","Engineering","Business","Medicine"],n),
    "Score": np.random.randint(50,100,n),
    "Attendance": np.random.randint(60,100,n)
})
```

```
lagos = df[df["City"]=="Lagos"]
print(f"Lagos students: {len(lagos)}")
high = df[(df["Score"]>=85)&(df["Attendance"]>=80)]
print(f"High performers: {len(high)}")
```

```
print(df.groupby("Department")["Score"].mean().round(2))
print(df.groupby(["City","Gender"]).size().unstack(fill_value=0))
print(df.groupby("Department").agg(
    Avg=("Score","mean"), Max=("Score","max"), Count=("Score","count")
).round(2))
```

```
df["Grade"] = df["Score"].apply(lambda x: "A" if x>=90 else "B" if x>=80 else "C" if x>=70 else "D")
df["Passed"] = df["Score"] >= 60
print(df[["Name","Score","Grade","Passed"]].head(10))
```
