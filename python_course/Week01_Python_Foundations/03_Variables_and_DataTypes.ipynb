## What is a Variable?
A variable is a **named container** that stores a value in memory.

```
name = "Kwanda"
age = 25
gpa = 9.25
is_student = True
print(name, age, gpa, is_student)
```

## Python Data Types
| Type | Example | Description |
|------|---------|-------------|
| `int` | `25` | Whole numbers |
| `float` | `9.25` | Decimal numbers |
| `str` | `"Hello"` | Text |
| `bool` | `True/False` | Logical values |
| `NoneType` | `None` | No value |

```
students = 30
print(type(students))

pi = 3.14159
print(type(pi))

course = "Python Masterclass"
print(type(course))

is_enrolled = True
print(type(is_enrolled))
```

## String Basics

```
name1 = "Muhammed"
print(len(name1))
print(name1[0])
print(name1[-1])
print(name1[0:4])
```

## Type Conversion

```
age_str = "25"
age_int = int(age_str)
age_float = float(age_str)
print(type(age_int))
print(type(age_float))

score = 95
score_str = str(score)
print(type(score_str))
```

## f-Strings (Modern String Formatting)

```
name = "Kwanda"
gpa = 9.25
print(f"My name is {name}")
print(f"My GPA is {gpa:.1f}")
print(f"2 + 2 = {2 + 2}")
```
