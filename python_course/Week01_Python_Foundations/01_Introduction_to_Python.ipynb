## What is Python?
Python is a high-level, general-purpose language created by **Guido van Rossum** in **1991**. It emphasises readability — you can express ideas in fewer lines than C++ or Java.

## A Brief History
| Version | Year | Highlights |
|---------|------|------------|
| Python 1.0 | 1994 | First release |
| Python 2.0 | 2000 | List comprehensions |
| Python 3.0 | 2008 | Unicode, major improvements |
| Python 3.11+ | 2022+ | Faster, better error messages |
