## Arithmetic Operators

```
a, b = 10, 3
print(f"Addition:       {a} + {b} = {a + b}")
print(f"Subtraction:    {a} - {b} = {a - b}")
print(f"Multiplication: {a} * {b} = {a * b}")
print(f"Division:       {a} / {b} = {a / b}")
print(f"Floor Division: {a} // {b} = {a // b}")
print(f"Modulus:        {a} % {b} = {a % b}")
print(f"Exponent:       {a} ** {b} = {a ** b}")
```

## Comparison Operators

```
x, y = 10, 20
print(x == y)
print(x != y)
print(x > y)
print(x < y)
print(x >= 10)
print(x <= 10)
```

## Logical Operators

```
age = 20
has_id = True
print(age >= 18 and has_id)
print(age >= 18 or has_id)
print(not has_id)
```

## String Operations

```
first = "Muhammed"
last = "Kwanda"
full = first + " " + last
print(full)
print(full.upper())
print(full.lower())
print(full.replace("Kwanda", "Musa"))
print(full.split(" "))
print("  hello  ".strip())
print(full.count("a"))
```
