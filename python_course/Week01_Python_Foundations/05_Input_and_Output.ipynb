## Getting User Input
`input()` always returns a **string**. Convert if needed.

```
name = input("What is your name? ")
print(f"Hello, {name}!")
```

```
age = int(input("Enter your age: "))
print(f"Next year you will be {age + 1}")

gpa = float(input("Enter your GPA: "))
print(f"Your GPA is {gpa:.2f}")
```

## Formatting Output

```
name = "Kwanda"
score = 95.678
print(f"Name: {name}")
print(f"Score: {score:.2f}")
print(f"Score: {score:.0f}")
print(f"Score: {score:10.2f}")
```

```
print(f"{"Name":<15} {"Score":>10} {"Grade":^10}")
print("-" * 35)
students = [("Amina", 92, "A"), ("Bello", 78, "B"), ("Chioma", 85, "B+")]
for n, s, g in students:
    print(f"{n:<15} {s:>10} {g:^10}")
```

## Simple Interactive Program

```
print("=== Simple Calculator ===")
num1 = float(input("First number: "))
op = input("Operator (+,-,*,/): ")
num2 = float(input("Second number: "))
if op == "+":
    result = num1 + num2
elif op == "-":
    result = num1 - num2
elif op == "*":
    result = num1 * num2
elif op == "/":
    result = num1 / num2 if num2 != 0 else "Error: div by zero"
else:
    result = "Invalid operator"
print(f"Result: {num1} {op} {num2} = {result}")
```
