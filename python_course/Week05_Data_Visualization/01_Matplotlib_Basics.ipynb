```
import matplotlib.pyplot as plt
import numpy as np
%matplotlib inline
```

```
months=["Jan","Feb","Mar","Apr","May","Jun","Jul","Aug","Sep","Oct","Nov","Dec"]
sales=[45,52,48,61,70,65,80,75,85,90,88,95]
plt.figure(figsize=(10,5))
plt.plot(months,sales,color="steelblue",linewidth=2,marker="o",markersize=5)
plt.title("Monthly Sales 2024",fontsize=14,fontweight="bold")
plt.xlabel("Month"); plt.ylabel("Sales (thousands)")
plt.grid(True,alpha=0.3); plt.tight_layout(); plt.show()
```

```
departments=["CS","Engineering","Business","Medicine","Law"]
students=[120,95,80,110,65]
plt.figure(figsize=(8,5))
bars=plt.bar(departments,students,color=["#2196F3","#4CAF50","#FF9800","#E91E63","#9C27B0"])
plt.title("Students per Department")
for bar in bars:
    plt.text(bar.get_x()+bar.get_width()/2, bar.get_height()+1, str(int(bar.get_height())), ha="center",fontweight="bold")
plt.tight_layout(); plt.show()
```

```
np.random.seed(42)
hours=np.random.uniform(1,10,50)
scores=hours*8+np.random.normal(0,5,50)
plt.figure(figsize=(8,5))
plt.scatter(hours,scores,color="tomato",alpha=0.7,edgecolors="white",s=80)
plt.title("Study Hours vs Score"); plt.xlabel("Hours"); plt.ylabel("Score")
plt.grid(True,alpha=0.3); plt.tight_layout(); plt.show()
```

```
fig,axes=plt.subplots(2,2,figsize=(12,8))
fig.suptitle("Student Dashboard",fontsize=16,fontweight="bold")
x=np.linspace(0,10,100)
axes[0,0].plot(x,np.sin(x)); axes[0,0].set_title("Trend")
axes[0,1].bar(["A","B","C","D","F"],[25,35,20,15,5],color="green"); axes[0,1].set_title("Grades")
axes[1,0].scatter(np.random.randn(50),np.random.randn(50),alpha=0.6); axes[1,0].set_title("Correlation")
axes[1,1].hist(np.random.normal(75,10,200),bins=20,color="purple",alpha=0.7); axes[1,1].set_title("Distribution")
plt.tight_layout(); plt.show()
```
