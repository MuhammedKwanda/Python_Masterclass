```
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
sns.set_theme(style="whitegrid")
%matplotlib inline

np.random.seed(42)
n=200
df=pd.DataFrame({
    "Gender": np.random.choice(["Male","Female"],n),
    "Department": np.random.choice(["CS","Engineering","Business","Medicine"],n),
    "Math": np.random.randint(50,100,n),
    "English": np.random.randint(50,100,n),
    "Science": np.random.randint(50,100,n),
    "Attendance": np.random.randint(60,100,n)
})
```

```
fig,axes=plt.subplots(1,2,figsize=(12,4))
sns.histplot(df["Math"],kde=True,ax=axes[0])
axes[0].set_title("Math Distribution")
sns.histplot(data=df,x="Math",hue="Gender",kde=True,ax=axes[1])
axes[1].set_title("Math by Gender")
plt.tight_layout(); plt.show()
```

```
fig,axes=plt.subplots(1,2,figsize=(14,5))
sns.boxplot(data=df,x="Department",y="Math",ax=axes[0],palette="Set2")
axes[0].set_title("Math by Department")
sns.violinplot(data=df,x="Department",y="Math",hue="Gender",split=True,ax=axes[1],palette="Set1")
axes[1].set_title("Math by Dept & Gender")
plt.tight_layout(); plt.show()
```

```
corr=df[["Math","English","Science","Attendance"]].corr()
plt.figure(figsize=(7,5))
sns.heatmap(corr,annot=True,fmt=".2f",cmap="coolwarm",center=0,square=True,linewidths=0.5)
plt.title("Correlation Matrix"); plt.tight_layout(); plt.show()
```

```
sns.pairplot(df[["Math","English","Science","Gender"]],hue="Gender",plot_kws={"alpha":0.5},palette="husl")
plt.suptitle("Pairplot of Scores",y=1.02); plt.show()
```
