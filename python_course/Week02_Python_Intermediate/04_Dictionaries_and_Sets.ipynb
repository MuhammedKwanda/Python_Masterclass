## Dictionaries

```
student = {
    "name": "Muhammed Kwanda",
    "age": 25,
    "gpa": 9.25,
    "university": "Sanskaram University",
    "courses": ["Network Security", "Cryptography"]
}
print(student["name"])
print(student.get("gpa"))
print(student.get("phone", "Not provided"))
```

```
print(student.keys())
print(student.values())
for key, value in student.items():
    print(f"{key}: {value}")
```

```
class_data = {
    "Amina": {"score": 92, "grade": "A"},
    "Bello": {"score": 78, "grade": "B"}
}
for name, info in class_data.items():
    print(f"{name}: {info['score']} ({info['grade']})")
```

## Sets

```
numbers = {1, 2, 3, 2, 1, 4}
print(numbers)

names = ["Amina","Bello","Amina","Chioma"]
print(set(names))
```

```
python_s = {"Amina","Bello","Chioma","Dauda"}
java_s = {"Bello","Efua","Chioma","Frank"}
print(python_s & java_s)
print(python_s | java_s)
print(python_s - java_s)
```
