## if / elif / else

```
temperature = 35
if temperature > 30:
    print("It's hot today!")
```

```
age = 17
if age >= 18:
    print("You can vote.")
else:
    print("You cannot vote yet.")
```

```
score = 75
if score >= 90: grade = "A"
elif score >= 80: grade = "B"
elif score >= 70: grade = "C"
elif score >= 60: grade = "D"
else: grade = "F"
print(f"Score: {score} -> Grade: {grade}")
```

## Nested Conditionals

```
age = 20
has_ticket = True
if age >= 18:
    if has_ticket:
        print("Welcome!")
    else:
        print("Need a ticket.")
else:
    print("Adults only.")
```

## Ternary Operator

```
age = 20
status = "Adult" if age >= 18 else "Minor"
print(status)
```

## Membership Operators

```
fruits = ["apple", "mango", "banana"]
print("mango" in fruits)
print("grape" not in fruits)

day = "Saturday"
if day in ["Saturday", "Sunday"]:
    print("Weekend!")
else:
    print("Weekday.")
```
