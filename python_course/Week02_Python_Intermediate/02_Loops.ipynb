## for Loop

```
for i in range(5):
    print(f"Iteration {i}")

for i in range(1, 11, 2):
    print(i)
```

```
students = ["Amina", "Bello", "Chioma", "Dauda"]
for student in students:
    print(f"Hello, {student}!")

for index, student in enumerate(students, start=1):
    print(f"{index}. {student}")
```

## while Loop

```
count = 1
while count <= 5:
    print(f"Count: {count}")
    count += 1
print("Done!")
```

## break, continue, pass

```
for i in range(10):
    if i == 5:
        print("Found 5! Stopping.")
        break
    print(i)

print("---")
for i in range(10):
    if i % 2 == 0:
        continue
    print(i)
```

## Nested Loops

```
for i in range(1, 6):
    for j in range(1, 6):
        print(f"{i*j:4}", end="")
    print()
```
