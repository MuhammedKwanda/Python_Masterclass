## Lists

```
fruits = ["apple", "mango", "banana", "orange"]
numbers = [1, 2, 3, 4, 5]
print(fruits)
print(type(fruits))
```

```
scores = [85, 92, 78, 95, 67, 88]
print(scores[0])
print(scores[-1])
print(scores[1:4])
print(scores[::-1])
```

```
names = ["Chioma", "Amina", "Bello"]
names.append("Dauda")
names.insert(1, "Fatima")
names.remove("Bello")
names.sort()
print(names)
print(len(names))
```

## Tuples — Immutable

```
coordinates = (10.5, 20.3)
print(coordinates[0])
x, y = coordinates
print(f"x={x}, y={y}")
a, b = 5, 10
a, b = b, a
print(f"a={a}, b={b}")
```
