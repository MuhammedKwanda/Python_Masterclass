## Basic Functions

```
def greet():
    print("Hello, welcome to Python Masterclass!")

greet()
```

```
def greet_student(name, course="Python"):
    print(f"Hello {name}! Welcome to {course} class.")

greet_student("Amina")
greet_student("Bello", "Machine Learning")
```

```
def add(a, b):
    return a + b

def calculate_grade(score):
    if score >= 90: return "A"
    elif score >= 80: return "B"
    elif score >= 70: return "C"
    elif score >= 60: return "D"
    else: return "F"

print(add(10, 25))
print(calculate_grade(85))
```

## *args and **kwargs

```
def sum_all(*numbers):
    return sum(numbers)

print(sum_all(1, 2, 3))
print(sum_all(10, 20, 30, 40))

def print_info(**details):
    for key, value in details.items():
        print(f"{key}: {value}")

print_info(name="Kwanda", age=25, city="Abuja")
```

## Lambda Functions

```
square = lambda x: x ** 2
print(square(5))

students = [("Amina", 92), ("Bello", 78), ("Chioma", 85)]
sorted_s = sorted(students, key=lambda s: s[1], reverse=True)
print(sorted_s)
```
