```
np.random.seed(42)
n=3000
normal=pd.DataFrame({"Flow_Duration":np.random.normal(50000,10000,int(n*0.8)),"Total_Fwd":np.random.normal(10,3,int(n*0.8)),"Total_Bwd":np.random.normal(8,2,int(n*0.8)),"Flow_Bytes_s":np.random.normal(5000,1000,int(n*0.8)),"Flow_Pkts_s":np.random.normal(100,20,int(n*0.8)),"Label":"BENIGN"})
attack=pd.DataFrame({"Flow_Duration":np.random.normal(1000,500,int(n*0.2)),"Total_Fwd":np.random.normal(50,10,int(n*0.2)),"Total_Bwd":np.random.normal(2,1,int(n*0.2)),"Flow_Bytes_s":np.random.normal(50000,10000,int(n*0.2)),"Flow_Pkts_s":np.random.normal(1000,200,int(n*0.2)),"Label":"ATTACK"})
df=pd.concat([normal,attack],ignore_index=True).sample(frac=1,random_state=42)
print(df.shape)
print(df["Label"].value_counts())
```
