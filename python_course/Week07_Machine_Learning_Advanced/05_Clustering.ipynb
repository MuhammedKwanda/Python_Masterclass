> Unsupervised learning — finding groups without labels.

```
import numpy as np, pandas as pd
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler
%matplotlib inline
```

```
np.random.seed(42)
customers=pd.DataFrame({"age":np.concatenate([np.random.normal(25,3,100),np.random.normal(45,5,100),np.random.normal(60,4,100)]),"spend":np.concatenate([np.random.normal(50000,10000,100),np.random.normal(150000,20000,100),np.random.normal(80000,15000,100)]),"freq":np.concatenate([np.random.normal(2,0.5,100),np.random.normal(8,1,100),np.random.normal(4,1,100)])})
plt.scatter(customers["age"],customers["spend"],alpha=0.5)
plt.xlabel("Age"); plt.ylabel("Annual Spend"); plt.title("Before Clustering"); plt.show()
```

```
scaler=StandardScaler()
X_s=scaler.fit_transform(customers)
inertias=[]
for k in range(1,11):
    km=KMeans(n_clusters=k,random_state=42,n_init=10)
    km.fit(X_s)
    inertias.append(km.inertia_)
plt.plot(range(1,11),inertias,"bo-"); plt.xlabel("k"); plt.ylabel("Inertia")
plt.title("Elbow Method"); plt.grid(True,alpha=0.3); plt.show()
```

```
km=KMeans(n_clusters=3,random_state=42,n_init=10)
customers["Segment"]=km.fit_predict(X_s)
colors=["red","blue","green"]
for seg,col in zip([0,1,2],colors):
    mask=customers["Segment"]==seg
    plt.scatter(customers.loc[mask,"age"],customers.loc[mask,"spend"],c=col,alpha=0.6,label=f"Seg {seg}")
plt.legend(); plt.xlabel("Age"); plt.ylabel("Spend"); plt.title("Customer Segments"); plt.show()
print(customers.groupby("Segment").mean().round(0))
```
