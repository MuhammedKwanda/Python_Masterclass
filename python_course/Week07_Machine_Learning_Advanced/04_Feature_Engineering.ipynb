```
import numpy as np, pandas as pd
from sklearn.preprocessing import StandardScaler, MinMaxScaler, LabelEncoder, OneHotEncoder
from sklearn.feature_selection import SelectKBest, f_classif
from sklearn.ensemble import RandomForestClassifier
import matplotlib.pyplot as plt
%matplotlib inline
```

```
df=pd.DataFrame({"city":["Lagos","Abuja","Lagos","Kano","Abuja"],"education":["BSc","MSc","PhD","BSc","MSc"],"salary":[150000,250000,350000,120000,220000],"promoted":[0,1,1,0,1]})
le=LabelEncoder()
df["edu_encoded"]=le.fit_transform(df["education"])
df_enc=pd.get_dummies(df,columns=["city"],drop_first=True)
print(df_enc)
```

```
data=np.array([[100,0.5],[200,1.2],[150,0.8],[300,2.1]])
scaler_std=StandardScaler()
scaler_mm=MinMaxScaler()
print("Standard:\n",scaler_std.fit_transform(data).round(3))
print("MinMax:\n",scaler_mm.fit_transform(data).round(3))
```

```
from sklearn.datasets import load_breast_cancer
bc=load_breast_cancer()
X=pd.DataFrame(bc.data,columns=bc.feature_names); y=bc.target
selector=SelectKBest(f_classif,k=10)
selector.fit(X,y)
print("Top 10 features:", list(X.columns[selector.get_support()]))

rf=RandomForestClassifier(n_estimators=100,random_state=42)
rf.fit(X,y)
pd.Series(rf.feature_importances_,index=X.columns).nlargest(10).plot(kind="barh",figsize=(8,5))
plt.title("Feature Importance"); plt.tight_layout(); plt.show()
```
