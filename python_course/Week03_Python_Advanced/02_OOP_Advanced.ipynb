## Inheritance

```
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age
    def introduce(self):
        return f"Hi, I'm {self.name}, age {self.age}"

class Student(Person):
    def __init__(self, name, age, student_id, gpa):
        super().__init__(name, age)
        self.student_id = student_id
        self.gpa = gpa
    def introduce(self):
        return f"Hi, I'm {self.name}, ID: {self.student_id}, GPA: {self.gpa}"

class Teacher(Person):
    def __init__(self, name, age, subject):
        super().__init__(name, age)
        self.subject = subject
    def introduce(self):
        return f"I'm {self.name}, I teach {self.subject}"

people = [Person("Ali",30), Student("Amina",20,"STU001",9.25), Teacher("Kwanda",35,"Cybersecurity")]
for p in people:
    print(p.introduce())
```

## Encapsulation

```
class SecureAccount:
    def __init__(self, owner, pin):
        self.owner = owner
        self.__pin = pin
        self.__balance = 0

    def deposit(self, amount, pin):
        if self.__verify(pin):
            self.__balance += amount
            print(f"Deposited. Balance: {self.__balance}")
        else:
            print("Wrong PIN!")

    def get_balance(self, pin):
        return self.__balance if self.__verify(pin) else "Access denied"

    def __verify(self, pin):
        return pin == self.__pin

acc = SecureAccount("Kwanda", 1234)
acc.deposit(5000, 1234)
acc.deposit(5000, 9999)
```
