```
with open("students.txt", "w") as f:
    f.write("Amina,92,A\n")
    f.write("Bello,78,B\n")
    f.write("Chioma,85,B+\n")

with open("students.txt", "r") as f:
    print(f.read())

with open("students.txt", "r") as f:
    for line in f:
        print(line.strip())
```

```
import csv
students = [["Name","Score","Grade"],["Amina",92,"A"],["Bello",78,"B"]]
with open("students.csv", "w", newline="") as f:
    writer = csv.writer(f)
    writer.writerows(students)

with open("students.csv", "r") as f:
    reader = csv.DictReader(f)
    for row in reader:
        print(f"{row['Name']}: {row['Score']} ({row['Grade']})")
```

```
import json
class_data = {
    "class": "Python Masterclass",
    "instructor": "Muhammed Kwanda",
    "students": [{"name": "Amina","score": 92}]
}
with open("class.json", "w") as f:
    json.dump(class_data, f, indent=2)

with open("class.json", "r") as f:
    loaded = json.load(f)
print(loaded["instructor"])
```
