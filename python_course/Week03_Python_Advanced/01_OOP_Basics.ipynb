## Classes & Objects

```
class Student:
    def __init__(self, name, age, gpa):
        self.name = name
        self.age = age
        self.gpa = gpa

    def greet(self):
        return f"Hi, I'm {self.name}, GPA: {self.gpa}"

s1 = Student("Amina", 20, 9.0)
s2 = Student("Bello", 22, 8.5)
print(s1.greet())
print(s2.greet())
```

```
class BankAccount:
    def __init__(self, owner, balance=0):
        self.owner = owner
        self.balance = balance
        self.transactions = []

    def deposit(self, amount):
        if amount > 0:
            self.balance += amount
            self.transactions.append(f"Deposit: +{amount}")
            print(f"Deposited {amount}. Balance: {self.balance}")

    def withdraw(self, amount):
        if amount > self.balance:
            print("Insufficient funds!")
        else:
            self.balance -= amount
            self.transactions.append(f"Withdrawal: -{amount}")
            print(f"Withdrew {amount}. Balance: {self.balance}")

    def statement(self):
        print(f"=== {self.owner} ===")
        for t in self.transactions:
            print(f"  {t}")
        print(f"  Balance: {self.balance}")

acc = BankAccount("Muhammed", 5000)
acc.deposit(2000)
acc.withdraw(1500)
acc.statement()
```

## Special Methods

```
class Student:
    def __init__(self, name, gpa):
        self.name = name
        self.gpa = gpa
    def __str__(self):
        return f"Student({self.name}, GPA={self.gpa})"
    def __gt__(self, other):
        return self.gpa > other.gpa

s1 = Student("Amina", 9.25)
s2 = Student("Bello", 8.75)
print(s1)
print(s1 > s2)
```
