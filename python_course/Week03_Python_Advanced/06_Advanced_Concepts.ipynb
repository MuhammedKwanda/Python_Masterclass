## List Comprehensions

```
squares = [x**2 for x in range(10)]
evens = [x for x in range(20) if x % 2 == 0]
print(squares)
print(evens)

matrix = [[i*j for j in range(1,4)] for i in range(1,4)]
for row in matrix:
    print(row)
```

```
students = {"Amina": 92, "Bello": 78, "Chioma": 85, "Dauda": 65}
passing = {name: score for name, score in students.items() if score >= 70}
print(passing)
```

## map(), filter(), zip()

```
scores = [78, 92, 85, 67, 95, 72]
doubled = list(map(lambda x: x*2, scores))
passing = list(filter(lambda x: x >= 70, scores))
print(doubled)
print(passing)

names = ["Amina","Bello","Chioma"]
marks = [92, 78, 85]
combined = list(zip(names, marks))
print(combined)
```

## Generators

```
def count_up(n):
    i = 0
    while i < n:
        yield i
        i += 1

for num in count_up(5):
    print(num)

gen_expr = (x**2 for x in range(10))
print(list(gen_expr))
```
